### Sequential Model 
โมเดลในที่นี้จะถูกสร้างขึ้นให้อยู่ในรูปแบบ sequential หมายถึงว่าการเรียงตัวแบบเป็นชั้นๆ แต่ละเลเยอร์จะส่งข้อมูลจากด้านใดไปด้านหนึ่งเท่านั้น (ซ้ายไปขวา)
โดยจะใส่เมธอด `Sequential()` ของ keras และนำlayer ใส่ไปใน list ต่อกันเป็นทอดๆ output ของ layer หนึ่งจะเป็น input ของอีก layer

การตั้งค่าการรับ input ของโมเดลสามารถทำได้ 2 วิธี
1. เป็นการกำหนดไปเลยว่าเลเยอร์แรกนั้นเป็นแบบ แบน(flatten) ***แต่ไม่กำหนดว่ามี input ขนาดเท่าไหร่***
   <br> ถ้าหากต้องการกำหนดรูปแบบ input ให้ใส่พารามิเตอร์ input_shape=(28,28) อาจจะจะขึ้น warning
   <br> warning เพราะ keras มองว่า input ไม่ควรเป็น layer เพราะเลเยอร์นี้ไม่ได้แก้ไขค่าอะไรเลยเป็นการส่งตัวแปรเข้าไปเท่านั้น activation fn ก็ไม่มี **แล้วก็ parameter นี้อาจจะ obsolete หากอัพเดท version ใหม่ๆ**
> Flatten layer คือเลเยอร์ที่เป็นเลเยอร์เหมือน Array 1D ยาวเป็นเส้นตรงแบนๆ
1. เป็นการกำหนดไปเลยว่า input จะเป็น array 2 มิติ ที่มีขนาด 28x28 = 784 pixel
    <br> แต่ต้องอย่าลืมปรับเปลี่ยน shape ของข้อมูลด้วย

จริงๆแล้วสองวิธีนี้ไม่มีผิดไม่มีถูก ขึ้นอยู่กับว่าถนัดแบบไหน

````python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)), 
    keras.layers.Dense(10, activation=tf.nn.softmax)
])

# model = keras.Sequential([
#     keras.Input(shape=(28,28)),
#     keras.layers.Flatten(),
#     keras.layers.Dense(10, activation=tf.nn.softmax)
# ])
````

จากโค้ดด้านบนจะได้ว่า โมเดลของเรามีการเรียงตัวแบบนี้
- `Flatten`: แปลงภาพ 28x28 เป็นอาเรย์ยาว 784 ค่า (ทำหน้าที่เป็น input layer)
- `Dense(10, softmax)`: ชั้นเอาต์พุตที่ให้ความน่าจะเป็น 10 คลาส (ทำหน้าที่เป็น output layer)

### Compile model
- ขั้นตอนนี้จะเป็นการบอกโมเดลว่าจะมีการใช้อัลกอริทีมอะไรในการ update weights
- คำนวน loss โดยการเปรียบเทียบแบบไหน (ขอติดไว้ก่อน)
- ใช้ metrics ใดในการแสดงประสิทธิภาพโมเดล

````python
model.compile(
    optimizer=keras.optimizers.SGD(),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
````

### Summary

ก่อนที่เราจะไปเริ่มฝึกโมเดลกัน ขั้นตอนที่ควรจะทำก่อนคือเรียก `summary()` ออกมาก่อนว่าโมเดลนั้นมีหน้าตา(เรียงตัว)เป็นอย่างไร
- บอก layer ว่าเป็นเลเยอร์ประเภทอะไร
- บอก output shape ด้วยว่าข้อมูลที่ออกมาจากเลเยอร์นั้นๆมี shape เป็นอย่างไร
- param ก็คือจำนวน weights ที่อยู่ในเลยอร์นั้นว่ามีจน.เท่าไหร่
- เช่น layer ประเภท `flatten` มี output ออกมา 784 (เพราะว่าเอา array2D 28x28 ไปทำเป็น 1D) แต่ไม่มี weights เพราะเป็น input layer

````python
model.summary()
````

### ฝึกฝนโมเดลด้วย `fit(...)`

คำสั่งนี้ใช้สำหรับ **ฝึกโมเดลด้วย train_images (ภาพ) และ train_labels (ป้ายชื่อหรือเฉลย)**

```python

```

โดยคำสั่งนี้จะ:
- ใช้ข้อมูลภาพจำนวนแค่ 1,000 ตัวอย่างแรก จาก `train_images` เพื่อทดสอบเบื้องต้นเฉยๆ
- ใช้ label ที่ตรงกัน จาก `train_labels`
- ฝึกซ้ำทั้งหมด 5 รอบ หรือ 5 epochs -> ภาพ 1000 วิ่งเข้าฝึกฝนโมเดลแค่ 5 รอบเท่านั้น

````python
model.fit(train_images[:1000],train_labels[:1000], epochs=5)
````

## Plot images with propability
- มิติใหม่แห่งการแสดงผลการทำนายคลาส เราจะแสดงระดับความน่าจะเป็นของแต่ละคลาสมาด้วย
- จะแสดงบาร์กราฟว่าแต่ละคลาสมีความน่าจะเป็นเท่าใด
- โดยจะบอกว่าโมเดลนั้นทายถูกหรือผิดด้วย ถ้าทายถูกจะเป็น<font color="blue">สีน้ำเงิน</font>/ทายผิดจะเป็น<font color="red">สีแดง</font>
- วิธีการนี้นอกจากจะช่วยให้เราเข้าใจผลลัพธ์ได้เร็ว มันยังช่วยสื่อสารให้คนเข้าใจได้ง่ายมากด้วย เหมาะกับการนำไปใส่ในรายงาน

อันดับแรก เราจะเริ่มจากการวาดรูปพร้อมกับลาเบลสีที่แสดงว่ามันทายถูกหรือผิดและทายคลาสที่มันตอบมาด้วยความน่าจะเป็นเท่าใด

โค้ดส่วนนี้จะเป็นตัวตัดสินในว่าจะพล็อต<font color="blue">สีน้ำเงิน</font> หรือ <font color="red">สีแดง</font>
``` python
predicted_label = np.argmax(predictions_array)
    if predicted_label == true_label:
        color = "blue"
    else:
        color = "red"

````python
def plot_image(i, predictions_array, true_label, img):
    predictions_array, true_label, img = predictions_array[i], true_label[i], img[i]
    plt.grid(False)
    plt.xticks([])
    plt.yticks([])

    plt.imshow(img, cmap=plt.cm.binary)

    predicted_label = np.argmax(predictions_array)
    if predicted_label == true_label:
        color = "blue"
    else:
        color = "red"

    plt.xlabel(
        "{} {:2.0f}% (true: {})".format(
            class_names[predicted_label],
            100 * np.max(predictions_array),
            class_names[true_label],
        ),
        color=color,
    )
````

เอาล่ะ คราวนี้มาดูฟังก์ชันสำหรับพล็อตบาร์กราฟกัน สามารถ copy ไปใช้กันได้

````python
def plot_value_array(i, predictions_array, true_label):
    predictions_array, true_label = predictions_array[i], true_label[i]
    plt.grid(False)
    # plt.xticks([])
    # plt.yticks([])
    thisplot = plt.bar(range(10), predictions_array, color="#777777") # plot bar
    plt.ylim([0, 1])
    predicted_label = np.argmax(predictions_array)
    # print(predicted_label)

    thisplot[predicted_label].set_color("red")
    thisplot[true_label].set_color("blue")
````

เราจะเรียกใช้ฟังก์ชันพล็อตทั้งสองนี้คู่กัน แน่นอนว่าด้วยคำสั่ง subplot ที่เราเรียนมาแล้วนั่นเอง และในที่นี้เราจะลองพล็อตตัวอย่างแรกออกมาดู

````python
i = 0 # index ของภาพ
plt.figure(figsize=(6, 3))
plt.subplot(1, 2, 1)
plot_image(i, predictions, test_labels, test_images)
plt.subplot(1, 2, 2)
plot_value_array(i, predictions, test_labels) #ความมั่นใจ

plt.tight_layout()
plt.show()
````

อันดับต่อไป เราจะลองมาดูตัวอย่างที่เครื่องทายผิด ภาพที่ 11

````python
i = 11
plt.figure(figsize=(6, 3))
plt.subplot(1, 2, 1)
plot_image(i, predictions, test_labels, test_images)
plt.subplot(1, 2, 2)
plot_value_array(i, predictions, test_labels)
````

### Multiple Plot
การดูภาพทีละภาพอาจใช้เวลานานและไม่สะดวก  
หากต้องการแสดงผล **หลายภาพพร้อมกันเป็นชุด** เราต้อง:

- จัดพื้นที่แสดงผลใหม่ (ใช้ `subplot()`)
- วนลูปพล็อตภาพและบาร์กราฟลงในพื้นที่ที่เตรียมไว้

---

ในตัวอย่างนี้ เราสามารถกำหนดได้ว่าอยากให้แสดงผลออกมากี่ **แถว** และกี่ **คอลัมน์**  
โดยระบุผ่านตัวแปร:

- `num_rows` → จำนวนแถวที่ต้องการแสดง
- `num_columns` → จำนวนคอลัมน์ (เฉพาะภาพ)

> จำนวนคอลัมน์ที่ใช้จริงในการจัดพื้นที่ **จะถูกคูณ 2**  
> เพราะการพล็อต 1 ตัวอย่าง = ภาพ 1 ภาพ + บาร์กราฟ 1 แผ่น  
> จึงต้องใช้ 2 ช่องต่อภาพ 1 ตัวอย่าง

การจัด layout แบบนี้จะทำให้เราสามารถดูภาพและผลการทำนายของโมเดลได้อย่างสะดวกและต่อเนื่อง


````python
num_rows = 5
num_cols = 5
num_images = num_rows * num_cols
plt.figure(figsize=(2 * 2 * num_cols, 2 * num_rows))
for i in range(num_images):
    plt.subplot(num_rows, 2 * num_cols, 2 * i + 1)
    plot_image(i, predictions, test_labels, test_images)
    plt.subplot(num_rows, 2 * num_cols, 2 * i + 2)
    plot_value_array(i, predictions, test_labels)
    # _ = plt.xticks(range(10), class_names, rotation=45)

plt.tight_layout()
````

## <font color="red">แบบฝึกหัดที่ 1 ผลของจำนวนข้อมูลชุดฝึก</font>
แบบฝึกหัดนี้จะช่วยเพิ่มความเข้าใจเกี่ยวกับปรับการเทรนโมเดล
ณ ตอนนี้หากรันโค้ดที่เตรียมไว้ให้จะ โมเดลจะมี acc อยู่ที่ประมาณ `60-70%`

แต่ตามทฤษฎีที่เรียนมาคือ โมเดลเราจะทำงานเก่งขึ้นก็ต่อเมื่อจำนวนข้อมูลที่เทรนนั้นมากขึ้น

**<font color="blue">มีข้อตกลงกันก่อนว่า ทุกครั้งที่ setting เปลี่ยน ให้นศ.รัน `sequential()` และ `compile()` ใหม่ทุกครั้ง(เตรียมไว้ให้แล้ว)เพื่อเป็นการ build โมเดลขึ้นมาใหม่ มิเช่นนั้นจะเป็นการฝึกทับไปที่โมเดลเดิม</font>**

ให้นศ.แก้ไขโค้ดให้ทำตาม setting ดังนี้ และรายงานด้วยว่าในแต่ละขั้นตอนการปรับนั้นทำให้โมเดลมีประสิทธิภาพ(acc & loss) เป็นเท่าไหร่:

ตัวอย่างการอ่านค่าประสิทธิภาพ 
```python
Epoch 5/5
32/32 ━━━━━━━━━━━━━━━━━━━━ 0s 2ms/step - accuracy: 0.7552 - loss: 0.7623
```
อ่านได้ว่าใน epoch รอบที่ 5 โมเดลมี accuracy 75% ดูจาก `accuracy: 0.7552` หรือจะก็อปไปวางเลยก็ได้ขอแค่ให้อ่านแบบ percent เป็น

1. ปรับ model.fit() ให้เทรนภาพ 10000 ภาพ epoch=5 รายงานผลจากรอบที่ 5
2. ปรับ model.fit() ให้เทรนภาพ 30000 ภาพ epoch=5 รายงานผลจากรอบที่ 5
3. ปรับ model.fit() ให้เทรนภาพ 50000 ภาพ epoch=5 รายงานผลจากรอบที่ 5
4. ปรับ model.fit() ให้เทรนภาพทั้งหมด 60000 ภาพ(โดยไม่ใส่เลข 60000) epoch=5 รายงานผลจากรอบที่ 5 


````python
# สำหรับแบบฝึกหัด 1
model = keras.Sequential([
    keras.Input(shape=(28,28)),
    keras.layers.Flatten(),
    keras.layers.Dense(10, activation=tf.nn.softmax)
])
model.compile(
    optimizer=keras.optimizers.SGD(),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
# model.summary()
````

````python
# สำหรับแบบฝึกหัด 1
model.fit(train_images[:],train_labels[:], epochs=5)
````

## <font color="red">แบบฝึกหัดที่ 2 อัตราการเรียนรู้ (Learning Rate)</font>
จริงๆแล้วการเทรนโมเดลเราสามารถปรับค่า learning rate ของโมเดลได้หากคิดว่าโมเดลนั้นมีการเรียนรู้ที่ช้า/เร็วเกินไป (**สังเกตุได้จากค่า acc และ loss ระหว่าง epoch ว่ามีการเปลี่ยนแปลงมากน้อยเพียงใด**)

โค้ดด้านล่างเป็นโค้ดที่ทำการ set ค่า learning rate ของ optimiser เอาโค้ดส่วนนี้ไปแทนการ complie โมเดลของเรา
```python
# Compile model ด้วย custom learning rate
learning_rate = 0.0001
optimizer = keras.optimizers.SGD(learning_rate=learning_rate)
model.compile(optimizer=optimizer, loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```
ให้นศ.แก้ไขโค้ด compile ให้ทำตาม setting ดังนี้จากนั้นทำการเทรนโมเดลด้วยภาพ 60000 ภาพ epoch=5 และรายงาน acc & loss ที่ epoch 1 และ epoch 2 ว่าเป็นเท่าไหร่:
1. ปรับ learning rate ของ SGD ให้เป็น 0.0001 (ตามโค้ดตัวอย่าง) 
2. ปรับ learning rate ของ SGD ให้เป็น 0.0005 
3. ปรับ learning rate ของ SGD ให้เป็น 0.001 
4. ใช้ optimiser เป็น Adam() และใช้ learning rate เป็น 0.0001 
   > ลองถาม ChatGPT ดูได้ว่าเรียกใช้อย่างไร
5. ใช้ optimiser เป็น Adam() และใช้ learning rate เป็น 0.001 

````python
# สำหรับแบบฝึกหัด 2
model = keras.Sequential([
    keras.Input(shape=(28,28)),
    keras.layers.Flatten(),
    keras.layers.Dense(10, activation=tf.nn.softmax)
])
# Compile model ด้วย custom learning rate
learning_rate = 0.001
optimizer = keras.optimizers.Adam(learning_rate = learning_rate)
model.compile(optimizer = optimizer, loss='sparse_categorical_crossentropy', metrics = ['accuracy'])

# model.summary()
````

````python
model.fit(train_images[:10000], train_labels[:10000], epochs = 5)
````

### <font color="red">แบบฝึกหัด 2.2 </font>

เราจะมาดูกันว่านอกจาก Accuracy ที่เพิ่มขึ้นในจังหวะฝึกแล้ว การทำนายกับ `test_images` ดีขึ้นหรือเปล่าโดยจะพล็อตรูปพร้อมกับค.น่าจะเป็นออกมา
1. ใช้โมเดลในแบบฝึกหัด 2 setting ที่ 5 และเรียก `prediction()` ดูโค้ดได้จาก[ตรงนี้](#Performance-Evaluation-on-Test-Images) และเก็บใส่ตัวแปร `predictions`
2. ทำการพล็อตภาพ 25 รูปแรก โดยให้มีรูปพร้อมกับกราฟค.น่าจะเป็น (น้ำเงิน/แดง) ออกมาคู่กันด้วย
3. เปรียบเทียบผลลัพธ์ระหว่างโมเดลเก่าที่ใช้ภาพแค่ 1000 ภาพและไม่มีการปรับ setting อะไร[จากกราฟนี้](#Multiple-Plot) กับโมเดลปัจจุบันที่ใช้ภาพมากกว่าและปรับไปแล้ว มีอะไรเปลี่ยนแปลงบ้าง
> accracy เพิ่ม/ลด? loss เพิ่ม/ลด? และ predictions ทายถูกมากขึ้น/ลดลง? (ตย.ภาพที่ทายได้ถูกแต่ก่อนหน้านี้ผิด)

````python
model = keras.Sequential([
    keras.Input(shape=(28,28)),
    keras.layers.Flatten(),
    keras.layers.Dense(10, activation=tf.nn.softmax)
])
# Compile model ด้วย custom learning rate
learning_rate = 0.001
optimizer = keras.optimizers.Adam(learning_rate = learning_rate)
model.compile(optimizer = optimizer, loss='sparse_categorical_crossentropy', metrics = ['accuracy'])

# model.summary()

model.fit(train_images[:1000], train_labels[:1000], epochs = 5)
predictions = model.predict(test_images)
````

````python
# สำหรับแบบฝึกหัด 2.2

num_rows = 5
num_cols = 5
num_images = num_rows * num_cols

plt.figure(figsize=(2 * 2 * num_cols, 2 * num_rows))
for i in range(num_images):
    plt.subplot(num_rows, 2 * num_cols, 2 * i + 1)
    plot_image(i, predictions, test_labels, test_images)
    plt.subplot(num_rows, 2 * num_cols, 2 * i + 2)
    plot_value_array(i, predictions, test_labels)
    # _ = plt.xticks(range(10), class_names, rotation=45)

plt.tight_layout()
plt.show()
````

## การแปลงผลการทายให้เป็นค่าความแม่นยำ <font color="red">(เกร็ดความรู้ อ่านภายหลังได้)</font>
ที่ผ่านมา เราสามารถตรวจสอบความถูกต้องของการทายตัวอย่างที่เราสนใจได้แล้ว อย่างไรก็ตาม โดยมากเราต้องการผลสรุปออกมาเป็นร้อยละของความแม่นยำ อันที่จริงเราจะวนลูปเทียบผลการทายทีละตัวระหว่างกับ test_labels แล้วนับจำนวนครั้งที่ทายถูกก็ได้

แต่การวนลูปพวกนี้มักจะไม่จำเป็น และทำให้การคำนวณช้าด้วย ในตอนนี้เรามีตัวอย่างแค่ระดับหมื่นตัวเราอาจจะยังไม่รู้สึกว่ามันช้า แต่โอกาศที่เราจะพบข้อมูลระดับแสนตัวก็นับว่ามีไม่น้อย ดังนั้นเราจะใช้วิธีอื่น ทำแบบ pythonic style

### ลูกเล่นของ np.argmax
เราอยากได้ลาเบลของคลาสที่เครื่องทายมาของแต่ละภาพ ซึ่งก็คืออินเด็กซ์ของคลาสที่มีความน่าจะเป็นสูงสุด หากเรารัน predictions.shape เราจะพบว่ามันมีขนาดเป็น (10000, 10) ซึ่งก็คือมีหนึ่งหมื่นตัวอย่าง แต่ละตัวอย่างมีผลการทาย 10 คลาส 

หากเรารัน `np.argmax(predictions)` มันจะวิ่งไปในอาเรย์ทั้งก้อนราวกับว่าอาเรย์ยาวต่อเนื่องเป็นแถวเดียว ไม่ได้เป็นอาเรย์สองมิติอย่างที่ควรเป็น และมันจะคืนค่าอินเด็กซ์ที่มีค่า max มาให้ค่าเดียว แทนที่จะคืนมาให้หมื่นค่าตามจำนวนตัวอย่าง

แต่ np.argmax อนุญาตให้เราระบุแกนมิติในอาเรย์ที่เราจะยึดถือเป็นแกนอ้างอิง เช่นถ้าเราบอกว่า np.argmax(predictions, axis=0) ทางนัมไพจะเอาแกนแรกซึ่งก็คือแถวมาเป็นแกนอ้างอิง นั่นคือในแถวมี 10 คอลัมน์ มันก็จะหาค่า max ในแต่ละคอลัมน์มาให้เป็นคำตอบ

ทว่าสิ่งที่เราต้องการคืออินเด็กซ์ค่า max ในแต่ละแถว ดังนั้นเราต้องสั่งให้ยึดคอลัมน์เป็นแกนอ้างอิง เพื่อที่มันจะได้หาว่าจากแนวตั้งซึ่งมี 10,000 แถว ในแต่ละแถวมีค่า max อยู่ที่อินเด็กซ์ใดบ้าง และนำไปสู่โค้ดที่เรียบง่ายว่า `np.argmax(predictions, axis=1)`

````python
predictions.shape
````

````python
predicted_classes = np.argmax(predictions, axis=1)
````

````python
predicted_classes[0:15]
````

**อย่าวนลูปเปรียบเทียบค่าระหว่าง NumPy array สองตัว แต่ให้ใช้ฟังก์ชัน np.equal แทน**
ในโค้ดตัวอย่างข้างล่างนี้ เราใช้ฟังก์ชันของนัมไพเปรียบเทียบค่าระหว่างอาเรย์สองตัว ซึ่งมันจะคืนผลลัพธ์มาเป็นอาเรย์อีกอันที่บันทึกว่าคู่ใดมีค่าเท่ากันคู่ใดมีค่าต่างกัน 

ในตัวอย่างนี้ เราเทียบค่าระหว่างอาเรย์ predictted_class และ test_labels และเก็บผลการเทียบไว้ในตัวแปรชื่อ correct ซึ่งหากช่องอาเรย์คู่ใดที่ค่าเท่ากัน ผลลัพธ์ที่อยู่ใน correct จะเป็น True แต่ช่องไหนที่ไม่เท่ากัน ผลลัพธ์ในอาเรย์ correct จะเป็น False

````python
correct = np.equal(predicted_classes, test_labels)
````

**ลองพิมพ์ค่าใน correct ออกมาดู** สังเกตด้วยว่าอาเรย์ในนัมไพก็เก็บค่าบูลีนได้ ไม่จำเป็นต้องเป็นตัวเลขแต่เพียงอย่างเดียว

````python
correct
````

### วิธีนับค่า True ในอาเรย์บูลีนตามวิถึ pythonic ไม่ควรอาศัยลูป
เวลาจะนับว่าข้างในอาเรย์มี True เราไม่ควรวนลูปแล้วเทียบทีละค่า แต่ให้ใช้ฟังก์ชัน sum ได้เลย ซึ่งไพธอนจะถือว่า True ในกระบวนการบวกมีค่าเป็น 1 และ False มีค่าเป็น 0

ดังนั้นหากเราจะคำนวณร้อยละความถูกต้อง เราสามารถนำจำนวน True ในอาเรย์ correct มาหารด้วยจำนวนตัวอย่างทดสอบทั้งหมด

### ค่าบูลีนในไพธอนรองรับการบวกลบ
ในการบวกลบ ไพธอนถือว่า True มีค่าเป็น 1 และ False มีค่าเป็น 0  ทดสอบดูกับเซลล์ข้างล่างนี้ได้เลย
